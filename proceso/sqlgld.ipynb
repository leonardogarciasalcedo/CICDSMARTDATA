```python
scope= 'accessScopeforADLS'
key= 'storageAccessKey'
```

```python
password =  dbutils.secrets.get(scope=f"{scope}",key=f"{key}")
```

```python
jdbcHostname = "serverevalua.database.windows.net"
jdbcDatabase = "dbgld"
jdbcPort = 1433
jdbcUsername = "evaluadmin"
jdbcPassword = f"{password}"
```

```python
# URL JDBC
jdbcUrl = f"jdbc:sqlserver://{jdbcHostname}:{jdbcPort};database={jdbcDatabase};encrypt=true;trustServerCertificate=false;hostNameInCertificate=*.database.windows.net;loginTimeout=30;"
```

```python
tableName = "Orders"

df = spark.read.format("jdbc") \
    .option("url", jdbcUrl) \
    .option("dbtable", tableName) \
    .option("user", jdbcUsername) \
    .option("password", jdbcPassword) \
    .load()
```

```python
df_query = spark.sql("""
    select order_id,user_id,Cust_Name,Cust_phone,Cust_Address || ' ' || Cust_Country as Cust_Address
 from dev.gold.product_cust_deliver
""")

# Stream or batch write the query results directly to JDBC
df_query.write \
  .format("jdbc") \
  .option("url", jdbcUrl) \
  .option("dbtable", tableName) \
  .option("user", jdbcUsername) \
  .option("password", jdbcPassword) \
  .mode("overwrite") \
  .save()
```
